```python
import numpy as np
import matplotlib.pyplot as plt
```

**Grafico simples com pontos azuis**


```python
x = np.arange(1.0,5.1,0.2)
y = np.exp(x)
plt.plot(x,y,'bo')
plt.title("Simple exponencial grapic")
plt.text(1,100, "Text in coordinates 1,100")
plt.show()
```

**Plote de várias funções simples**

```python
x = np.arange(1.0,5.1,0.2)
y = 1.2 ** x

x2 = x[:]
y2 = np.log(x2)

x3 = x
y3 = np.cos(x3)

plt.title("Three different functions")
plt.plot(x,y,'bo',x2,y2,x3,y3,'r--')

plt.legend(('exponential', 'logarithm','cosine'), loc = 'lower right')
plt.show()
```

**Utilizando subplots**

```python
plt.subplot(1,3,1)
pi = np.pi

x1 = np.arange(-4*pi,4*pi,0.1)
y1 = np.sin(x1)

plt.title('sin')
plt.plot(x1,y1,"b--")

plt.subplot(1,3,2)

x2 = x1
y2 = np.cos(x2)
plt.plot(x2,y2,"r--")
plt.title("Cosine")

plt.subplot(1,3,3)

x3 = x1
y3 = np.tan(x1)
plt.plot(x3,y3,'r')
plt.title("tangent")

plt.show()




```

**Salvando gráfico em png**

```python
fig = plt.figure()
x = np.arange(-10,10.1,0.1)
y = x**3 + 4*x + 2

plt.plot(x,y)

#aqui define até onde vai os eixos
plt.axis([-12,15,-1000,1000])

plt.xlabel('my axis X')
plt.ylabel('\nMy axis y')

fig.savefig( R'grafico.png')

plt.show()
```

**Plot de gráfico em barras**


```python
left = [1,2,3,4]
right = [42,21,16,4]

states = ['SP','MG','RJ','ES']

plt.bar(left,right,color = 'blue', tick_label = states)

plt.title("Pops from southeast states ")
plt.xlabel('States')
plt.ylabel("Million inhabitants")
```

**Criando histograma**

```python
mu = 100
sigma = 15

x = mu + sigma * np.random.randn(10000)
n, bins, patches = plt.hist(x,50)

plt.xlabel('Intelligency')
plt.ylabel('Probabolity')
plt.title('IQ Histogram')

plt.grid(True)
plt.show()
```
